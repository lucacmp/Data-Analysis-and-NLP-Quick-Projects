### IMPORTAZIONE DELLE LIBRERIE

```python
import pandas as pd
from sklearn.datasets import load_diabetes, load_wine, load_breast_cancer
from sklearn.decomposition import PCA, TruncatedSVD
import numpy as np
from scipy.stats import spearmanr, kendalltau
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
```

### FUNZIONI

```python
def compute_correlation(data, method='pearson'):
    """Calcola la correlazione tra le caratteristiche in base al metodo specificato e stampa le tre coppie più correlate."""

    correlation_matrix = data.corr(method=method) # calcolo la matrice di correlazione usando il metodo specificato

    mask = np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool) # creo una maschera che tiene conto solamente del triangolo superiore della matrice di correlazione.
                                                                        # 'np.triu' crea la maschera; 'np.ones' genera una matrice in cui tutti i valori sono 1 (con le stesse coordinate del dataset);
                                                                        # 'k=1' esclude i valori della diagonale dominante; .'.astype(bool)' sostituisce valori booleani agli 1/0

    filtered_correlations = correlation_matrix.where(mask) # filtro le correlazioni in base alla maschera che ho creato, utilizzando il metodo '.where'

    correlations = filtered_correlations.unstack().dropna() # trasformo la matrice in una serie (indici: valori) e butto i valori Null

    sorted_correlations = correlations.reindex(correlations.abs().sort_values(ascending=False).index) # metto in ordine le correlazioni sulla base della loro forza assoluta
                                                                                                      # '.abs' serve a non tenere conto del segno positivo/neg

    top_pairs = sorted_correlations.head(3) # prendo le prime 3 correlazioni

    print(f"Top 3 {method.capitalize()} correlated feature:")
    for (feat1, feat2), corr_value in top_pairs.items():
        print(f"{feat1} - {feat2}: {corr_value}")
    print("\n")

    return top_pairs
```

```python
def compute_pca(data, n_components=2, output_file="pca_reduced_data.csv"):
    """Esegue la PCA sul dataset, riduce a n_components e salva i risultati in un file CSV."""

    pca = PCA(n_components=n_components) # istanzio la PCA con il numero di componenti principali a cui desidero ridurre il df

    principal_components = pca.fit_transform(data)

    reduced_df = pd.DataFrame(principal_components, columns=[f"PC{i + 1}" for i in range(n_components)])

    reduced_df.to_csv(output_file, index=False) # salvo il df ridotto in un file csv

    print(f"PCA completata. Dati ridotti salvati in '{output_file}'.") # effettuo un print di controllo
    print("\n")
    return reduced_df
```

```python
def compute_svd(data, n_components=2, output_file="svd_reduced_data.csv"):
    """Esegue la SVD sul dataset, riduce a n_components e salva i risultati in un file CSV."""

    svd = TruncatedSVD(n_components=n_components)

    components = svd.fit_transform(data)

    reduced_df = pd.DataFrame(components, columns=[f"SVD{i + 1}" for i in range(n_components)])

    reduced_df.to_csv(output_file, index=False)

    print(f"SVD completata. Dati ridotti salvati in '{output_file}'.")
    print("\n")
    return reduced_df
```

```python
def plot_correlations(data, top_pairs):
    """Crea uno scatter plot con le coppie di feature più correlate."""

    for (feat1, feat2), corr_value in top_pairs.items(): # itero su ciascun elemento di top_pairs

        plt.figure(figsize=(8, 6))
        plt.scatter(data[feat1], data[feat2], alpha=0.6)
        plt.title(f'Scatter plot di {feat2} vs {feat1}\nCorrelazione: {corr_value:.2f}')
        plt.xlabel(feat1)
        plt.ylabel(feat2)
        plt.grid(True)

        plt.show()
```

### Diabetes analysis

```python
diabetes = load_diabetes()
```

Le colonne del dataset sono rinominate manualmente per rendere la lettura più comprensibile.

```python
diabetes_df = pd.DataFrame(data = diabetes.data, columns = ['age', 'sex', 'bmi', 'bp', 'tot serum chol', 'l-den lipro', 'h-den lipro', 'tot chol', 'ltg', 'blood sug lev'])
print(diabetes_df.head())
```

Calcolo della correlazione con metodo Pearson

```python
diabetes_pearson = compute_correlation(diabetes_df)

diabetes_top_pearson = diabetes_pearson.head(3)

plot_correlations(diabetes_df, diabetes_top_pearson)
```

```python
diabetes_pca = compute_pca(diabetes_df, n_components = 2, output_file='outputs/pca_reduced_diabetes.csv')
diabetes_svd = compute_svd(diabetes_df, n_components = 2, output_file='outputs/svd_reduced_diabetes.csv')

print(diabetes_pca.head())
print("\n")
print(diabetes_svd.head())
```

### Wine analysis

I dati di questo dataset hanno scale differenti per ciascuna feature, dunque è necessario svolgere una standardizzazione prima di procedere con il calcolo della correlazione, PCA ed SVD.

```python
scaler = StandardScaler()
wine = load_wine()
```

```python
wine_df = pd.DataFrame(data = wine.data, columns = wine.feature_names)
print(wine_df.head())
```

```python
wine_scaled = scaler.fit_transform(wine_df)
wine_scaled_df = pd.DataFrame(data = wine_scaled, columns = wine_df.columns)
print(wine_scaled_df.head())
```

Calcolo della correlazione con metodo Spearman

```python
wine_spearman = compute_correlation(wine_scaled_df, method='spearman')

wine_top_spearman = wine_spearman.head(3)

plot_correlations(wine_scaled_df, wine_top_spearman)
```

```python
wine_pca = compute_pca(wine_scaled_df, n_components = 2, output_file='outputs/pca_reduced_wine.csv')
wine_svd = compute_svd(wine_scaled_df, n_components = 2, output_file='outputs/svd_reduced_wine.csv')

print(wine_pca.head())
print("\n")
print(wine_svd.head())
```

### Breast cancer analysis

```python
cancer = load_breast_cancer()
```

```python
cancer_df = pd.DataFrame(data = cancer.data, columns = cancer.feature_names)
print(cancer_df.head())
```

```python
cancer_scaled = scaler.fit_transform(cancer_df)
cancer_scaled_df = pd.DataFrame(data = cancer_scaled, columns = cancer.feature_names)
print(cancer_scaled_df.head())
```

Calcolo della correlazione con metodo Kendall

```python
cancer_kendall = compute_correlation(cancer_scaled_df, method='kendall')

cancer_top_kendall = cancer_kendall.head(3)

plot_correlations(cancer_scaled_df, cancer_top_kendall)
```

```python
cancer_pca = compute_pca(cancer_scaled_df, n_components = 2, output_file='outputs/pca_reduced_cancer.csv')
cancer_svd = compute_svd(cancer_scaled_df, n_components = 2, output_file='outputs/svd_reduced_cancer.csv')

print(cancer_pca.head())
print("\n")
print(cancer_svd.head())
```
